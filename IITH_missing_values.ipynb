```python
students = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Branch": ["CSE", "ECE", "CSE", "IT", "CSE", "ECE"]
})
```

```python
students
```

```python
#Write a code to count the number of students within each department.
```

```python
students["Branch"].count()
```

```python
students["Branch"].value_counts()
```

```python
counts = {}
for branch in students['Branch']:
    if branch in counts:
        counts[branch] += 1
    else:
        counts[branch] = 1

print(counts)
```

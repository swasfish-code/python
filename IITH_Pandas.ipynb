```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Rahul", "Priya", "Amit", "Sneha", "Rohan",
        "Neha", "Arjun", "Pooja", "Karan", "Meera"
    ],
    "Branch": [
        "CSE", "ECE", "CSE", "IT", "CSE",
        "ECE", "IT", "CSE", "ECE", "IT"
    ],
    "Marks": [85, 78, 92, 67, 88, 76, 95, 81, 72, 89],
    "Attendance": [92, 85, 78, 65, 90, 88, 96, 72, 80, 94],
    "Study_Hours": [5, 4, 7, 2, 6, 4, 8, 5, 3, 6],
    "Placement_Training": [
        "Yes", "No", "Yes", "No", "Yes",
        "Yes", "Yes", "No", "No", "Yes"
    ]
})

students
```

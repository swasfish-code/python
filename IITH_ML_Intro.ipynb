#Rule-based system


```python
messages = [
    "Win a free prize now",
    "Meeting scheduled for tomorrow",
    "Claim your lottery reward",
    "Please send the project report",
    "You won a prize",
    "Can we meet after lunch?"
]

spam_words = ["win", "prize", "lottery", "claim"]

for message in messages:
    message_lower = message.lower()

    if any(word in message_lower for word in spam_words):
        print(message, "-> Possible spam")
    else:
        print(message, "-> Probably not spam")
```

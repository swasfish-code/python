```python
def prob(fav, out):
  f = len(fav)
  n = len(out)
  return round(f/n,2)
```

```python
Suppose a class has 100 students.

40 students are from CSE
30 students are from CSE and know Python

We want:

Probability that a student knows Python given that they are from CSE.

A ==> CS
B ==> Python

P(A) = 40/100
P(A^B) = 30/100

P(B|A) = P(A^B)/P(A) = 0.3/0.4 = 3/4 = 75%
```

```python
Take a standard deck of 52 cards.

There are:

26 red cards
26 black cards

13 hearts
13 diamonds
13 clubs
13 spades

Question:

What is the probability of selecting a heart?
13/52 = 0.25


What is the probability of selecting a Red card?
26/52 = 1/2


What is the probability of selecting a heart given that the card is red?

13/26 = 1/2
```

```python
A class has 50 students:

30 students know Python
20 students do not know Python
18 Python students are from CSE
12 non-Python students are from CSE

If a student is selected randomly, what is the probability that the student knows Python given that the student is from CSE?

A. 0.50
B. 0.60
C. 0.64
D. 0.72

A -> Python
B -> CSE

P(a) = 30/50
p(-a) = 20/50
p(a^b) = 18/50
P(-a^b) = 12/50
p(b) = P(a^b) + p(-a^b) = 18+12/50 = 30/50

p(a/b) = p(a^b)/p(b) = 18/30


18/30
```
